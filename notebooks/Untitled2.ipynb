```python
def compress_runs(lst):
    if not lst:
        return []
    result = []
    current = lst[0]
    count = 1

    for x in lst[1:]:
        if x == current:
            count += 1
        else:
            result.append([current, count])
            current = x
            count = 1

    result.append([current, count])
    return result
```

```python
print(compress_runs([1,1,1,2,3,3]))
```
