```python
# ==========================
# Import CSV
# ==========================

filepath = "../data/spike_waveforms_2.csv"


df = pd.read_csv(

    filepath,

    header=None

)


print(

    "Shape:",

    df.shape

)


display(
    df.head()
)
```

```python
# Confirm dimensions

n_timepoints = df.shape[0]
n_units = df.shape[1]

print("Time samples:", n_timepoints)
print("Units:", n_units)
```

```python
# Check data types
display(df.dtypes)
```

```python
# Summary statistics
display(df.describe())
```

```python
# Check NaNs
total_nan = df.isna().sum().sum()
print("Total NaN values:", total_nan)
nan_per_unit = df.isna().sum()
display(nan_per_unit[nan_per_unit > 0])
```

```python
# Check zeros
zero_total = (df == 0).sum().sum()
print("Total zeros:", zero_total)
zeros_per_unit = (df == 0).sum()
display(zeros_per_unit)
```

```python
# Plot random neuron traces
plt.figure(figsize=(14, 8))
random_rows = np.random.choice(
    df.index,
    10,
    replace=False
)

for row in random_rows:
    plt.plot(
        df.columns,
        df.iloc[row,:],
        alpha=0.7,
        label=f"Row {row}"
    )

plt.xlabel("Unit (column)")
plt.ylabel("Response")
plt.title("Random rows across units")
plt.legend()
plt.show()
```

```python
# Plot all neuron traces
plt.figure(figsize=(14, 8))

for col in df.columns:
    plt.plot(
        df.index,
        df[col],
        alpha=0.15
    )

plt.xlabel("Time sample")
plt.ylabel("Response")
plt.title("All neuron traces")
plt.show()
```

```python
# Heatmap
plt.figure(figsize=(16, 8))
sns.heatmap(
    df.T,
    cmap="coolwarm",
    center=0,
    cbar=True
)

plt.xlabel("Time sample")
plt.ylabel("Neuron / Unit")
plt.title("Neuron activity heatmap")
plt.show()
```

```python
# Histogram of all values
plt.figure(figsize=(10, 6))
plt.hist(
    df.values.flatten(),
    bins=100
)

plt.xlabel("Response")
plt.ylabel("Count")
plt.title("Distribution of all values")
plt.show()
```

```python
# Clean data
df_clean = df.interpolate(
    axis=0,
    limit_direction="both"
)

print("Original shape:", df.shape)
print("Clean shape:", df_clean.shape)
print("Remaining NaNs:", df_clean.isna().sum().sum())
```

```python
# Remove inactive all-zero units
inactive_units = df_clean.columns[
    (df_clean == 0).all()
    ]
print("Inactive all-zero units:")
print(list(inactive_units))

df_clean = df_clean.drop(
    columns=inactive_units
)

print("Shape after removing inactive units:", df_clean.shape)
```

```python
# Save cleaned CSV
output_path = "../data/spike_waveforms_clean.csv"
df_clean.to_csv(
    output_path,
    index=False,
    header=False
)

print("Saved cleaned data to:")
print(output_path)
```
